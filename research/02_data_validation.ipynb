```python
import os
import sys
import logging
from textSummarizer import logger
from textSummarizer.exception import CustomException
from textSummarizer.entity import DataValidationConfig
from textSummarizer.config.configuration import ConfigurationManager
```

```python
config_manager = ConfigurationManager()
data_validation_config = config_manager.get_data_validation_config()
print(data_validation_config)
```

```python
class DataValidation:
    def __init__(self, config: DataValidationConfig) -> DataValidationConfig:
        self.config = config

    def validate_all_files_exist(self):
        try:
            
            all_files = os.listdir(self.config.data_path)
            missing_files = [
                f for f in self.config.ALL_REQUIRED_FILES if f not in all_files
            ]
            if missing_files:
                validate_status = False
            else:
                validate_status = True
            with open(self.config.STATUS_FILE, "w") as f:
                f.write(f"Validate status : {validate_status}")
            return validate_status  

        except Exception as e:
            raise CustomException(e,sys)
```

```python
data_validation = DataValidation(config=data_validation_config)
result = data_validation.validate_all_files_exist()
print(result)
```
